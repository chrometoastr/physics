## Derivation of Electro-Magnetic Stress Energy Tensor from Electro-Magnetic Tensor

Question: How can I get the Stress-Energy tensor from the electromagnetic tensor?

You get the electromagnetic stress-energy tensor by forming a **quadratic combination** of the electromagnetic field tensor $F_{\mu\nu}$ with itself, in a way that produces:

* the correct energy density,
* the correct momentum/energy flux,
* the correct stresses,
* and a conserved tensor in vacuum.

In SI units, the standard result is

$$
T^{\mu\nu} = \frac{1}{\mu_0}
\left(
F^{\mu\alpha}F^\nu{}_{\alpha}
-\frac14 g^{\mu\nu}F_{\alpha\beta}F^{\alpha\beta}
\right).
$$

In Gaussian or Heaviside-Lorentz units, the prefactors look different, but the structure is the same.

This is exactly the EM version of the “stress-energy tensor” item in your roadmap. 

---

## 1. What the ingredients are

The electromagnetic field tensor is

$$
F_{\mu\nu}=
\partial_\mu A_\nu-\partial_\nu A_\mu,
$$

and in Cartesian inertial coordinates it packages $\mathbf E$ and $\mathbf B$ together.

A common convention is

$$
F^{0i}=\frac{E^i}{c},
\qquad
F^{ij}=-\epsilon^{ijk}B_k
$$

up to sign conventions depending on metric signature and index placement.

The point is:

* $F$ contains the electric and magnetic fields
* $T^{\mu\nu}$ is built from **products of $F$**

That makes sense physically because field energy density is quadratic in the fields:
$$
u_{\text{EM}} \sim E^2 + B^2.
$$

---

## 2. Why the formula has that shape

A rank-2 tensor must come from contracting indices appropriately. The natural quadratic object is

$$
F^{\mu\alpha}F^\nu{}_{\alpha}.
$$

But this by itself is not quite the right tensor. You also need to subtract off a trace term:

$$
-\frac14 g^{\mu\nu}F_{\alpha\beta}F^{\alpha\beta}.
$$

That second term ensures the final tensor has the correct pressure/stress behavior and, in 4D vacuum electromagnetism, is traceless.

So the final combination is

$$
T^{\mu\nu} = \frac{1}{\mu_0}
\left(
F^{\mu\alpha}F^\nu{}_{\alpha}
-\frac14 g^{\mu\nu}F_{\alpha\beta}F^{\alpha\beta}
\right).
$$

---

## 3. What each part means

## First term

$$
F^{\mu\alpha}F^\nu{}_{\alpha}
$$

This is the main quadratic “field times field” contribution.

It generates:

* energy density,
* momentum density,
* energy flux,
* stresses.

## Second term

$$
-\frac14 g^{\mu\nu}F_{\alpha\beta}F^{\alpha\beta}
$$

This adjusts the diagonal part properly.

Without it, you do not get the right Maxwell stress tensor structure.

---

## 4. The result in block form

When you unpack $T^{\mu\nu}$, you recover familiar classical EM quantities.

## Energy density

$$
T^{00} = u_{\text{EM}}
$$
with
$$
u_{\text{EM}} = \frac12\left(\epsilon_0 E^2+\frac{B^2}{\mu_0}\right)
$$
in SI units.

## Energy flux / momentum density

$$
T^{0i}
$$
contains the Poynting vector:
$$
\mathbf S = \frac{1}{\mu_0}\mathbf E\times \mathbf B.
$$

## Spatial stress

$$
T^{ij}
$$
is the Maxwell stress tensor.

So the field tensor $F_{\mu\nu}$ packages the fields, and the stress-energy tensor built from it packages:

* field energy,
* field momentum,
* field pressure/tension,
* field shear stress.

---

## 5. A concrete derivation sketch

Start from the electromagnetic Lagrangian density

$$
\mathcal L = -\frac{1}{4\mu_0}F_{\alpha\beta}F^{\alpha\beta}.
$$

Then vary the action with respect to the metric $g_{\mu\nu}$. The general rule is

$$
T^{\mu\nu} = \frac{2}{\sqrt{-g}}
\frac{\delta (\sqrt{-g},\mathcal L)}{\delta g_{\mu\nu}}.
$$

Applying that to the EM Lagrangian gives

$$
T^{\mu\nu} = \frac{1}{\mu_0}
\left(
F^{\mu\alpha}F^\nu{}_{\alpha}
-\frac14 g^{\mu\nu}F_{\alpha\beta}F^{\alpha\beta}
\right).
$$

So this is not just guessed: it comes directly from the action principle.

That connects nicely with your GR study emphasis on action principles, Noether ideas, and stress-energy. 

---

## 6. Useful invariant inside the formula

A key scalar appearing in the trace term is

$$
F_{\alpha\beta}F^{\alpha\beta}.
$$

In standard conventions this becomes

$$
F_{\alpha\beta}F^{\alpha\beta} = 2\left(B^2-\frac{E^2}{c^2}\right)
$$

up to sign convention.

So the trace term is built from the Lorentz-invariant combination of electric and magnetic fields.

---

## 7. Example: pure electric field in the $x$-direction

Suppose

$$
\mathbf E = (E,0,0), \qquad \mathbf B=0.
$$

Then the EM energy density is

$$
u = \frac12 \epsilon_0 E^2.
$$

There is no Poynting flux because $\mathbf E\times \mathbf B=0$.

The spatial part gives the field stresses:

* tension along the field direction,
* pressure in transverse directions,

depending on sign convention.

So even a static electric field has a nontrivial $T^{\mu\nu}$.

---

## 8. Example: plane electromagnetic wave

For a plane wave moving in $+z$,

* $T^{00}\neq 0$,
* $T^{03}\neq 0$,
* $T^{30}\neq 0$,
* $T^{33}\neq 0$.

This is a beautiful case because the wave carries:

* energy density,
* energy flux,
* momentum density,
* radiation pressure.

All of that comes from the same formula built out of $F_{\mu\nu}$.

---

\text{metric}\times \text{scalar field invariant}
}
$$

more explicitly,

$$
\boxed{
T^{\mu\nu} = \frac{1}{\mu_0}
\left(
F^{\mu\alpha}F^\nu{}_{\alpha}
-\frac14 g^{\mu\nu}F_{\alpha\beta}F^{\alpha\beta}
\right)
}
$$

---

##  10. Bottom line

To get the electromagnetic stress-energy tensor from the electromagnetic field tensor:

1. raise indices on $F_{\mu\nu}$ using the metric,
2. form the quadratic term $F^{\mu\alpha}F^\nu{}_{\alpha}$,
3. subtract the trace piece $\frac14 g^{\mu\nu}F_{\alpha\beta}F^{\alpha\beta}$,
4. include the unit-dependent prefactor.

That gives the tensor encoding EM:

* energy density,
* momentum density,
* energy flux,
* stresses.
