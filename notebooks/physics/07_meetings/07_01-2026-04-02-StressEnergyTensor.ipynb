## Electromagnetic Tensor

In standard relativistic notation, there is **not just one** Maxwell equation. There are **two tensor equations**:

$$
\partial_\mu F^{\mu\nu}=\mu_0 J^\nu
$$

and

$$
\partial_{[\alpha}F_{\beta\gamma]}=0
$$

equivalently,

$$
\partial_\mu \tilde F^{\mu\nu}=0.
$$

Together, these are the compact relativistic form of the usual **four Maxwell equations**.

Since you asked for the “single Maxwell equation,” the most likely one you mean is the first one,

$$
\partial_\mu F^{\mu\nu}=\mu_0 J^\nu,
$$

but that only gives **two** of the four classical equations:

* Gauss’s law
* Ampère–Maxwell law

The other two come from the second tensor equation.

---

# 1. Ingredients

The electromagnetic field tensor is

$$
F^{\mu\nu}=
\begin{pmatrix}
0 & -E_x/c & -E_y/c & -E_z/c\\
E_x/c & 0 & -B_z & B_y\\
E_y/c & B_z & 0 & -B_x\\
E_z/c & -B_y & B_x & 0
\end{pmatrix}
$$

up to sign convention. I will use coordinates

$$
x^\mu = (ct,x,y,z),
\qquad
\partial_\mu=\left(\frac{\partial}{\partial (ct)},\frac{\partial}{\partial x},\frac{\partial}{\partial y},\frac{\partial}{\partial z}\right).
$$

Also,

$$
J^\nu=(c\rho, J_x, J_y, J_z).
$$

---

# 2. The inhomogeneous tensor equation

$$
\partial_\mu F^{\mu\nu}=\mu_0 J^\nu
$$

This is really four component equations, one for each value of $\nu$.

## Case $\nu=0$

Then

$$
\partial_\mu F^{\mu 0}=\mu_0 J^0=\mu_0 c\rho.
$$

Expand the sum over $\mu$:

$$
\partial_0 F^{00}+\partial_1 F^{10}+\partial_2 F^{20}+\partial_3 F^{30}=\mu_0 c\rho.
$$

Since $F^{00}=0$,

$$
\partial_1 F^{10}+\partial_2 F^{20}+\partial_3 F^{30}=\mu_0 c\rho.
$$

Using

$$
F^{10}=E_x/c,\quad F^{20}=E_y/c,\quad F^{30}=E_z/c,
$$

we get

$$
\frac{1}{c}\left(\frac{\partial E_x}{\partial x}+\frac{\partial E_y}{\partial y}+\frac{\partial E_z}{\partial z}\right)=\mu_0 c\rho.
$$

So

$$
\nabla\cdot \mathbf E=\mu_0 c^2 \rho.
$$

Since

$$
\mu_0 c^2=\frac{1}{\epsilon_0},
$$

this becomes

$$
\boxed{\nabla\cdot \mathbf E=\frac{\rho}{\epsilon_0}}
$$

which is **Gauss’s law**.

---

## Case $\nu=i$ where $i=1,2,3$

Take $\nu=1$ first:

$$
\partial_\mu F^{\mu 1}=\mu_0 J^1=\mu_0 J_x.
$$

Expand:

$$
\partial_0 F^{01}+\partial_2 F^{21}+\partial_3 F^{31}=\mu_0 J_x.
$$

Using

$$
F^{01}=-E_x/c,\quad F^{21}=B_z,\quad F^{31}=-B_y,
$$

we get

$$
\frac{\partial}{\partial (ct)}\left(-\frac{E_x}{c}\right)
+\frac{\partial B_z}{\partial y}
-\frac{\partial B_y}{\partial z}
=\mu_0 J_x.
$$

Since $\partial/\partial(ct)=\frac{1}{c}\partial/\partial t$,

$$
-\frac{1}{c^2}\frac{\partial E_x}{\partial t}
+\frac{\partial B_z}{\partial y}
-\frac{\partial B_y}{\partial z}
=\mu_0 J_x.
$$

So

$$
(\nabla\times\mathbf B)_x-\frac{1}{c^2}\frac{\partial E_x}{\partial t}=\mu_0 J_x.
$$

Doing the same for $\nu=2,3$ gives the $y$- and $z$-components, so altogether:

$$
\boxed{\nabla\times\mathbf B-\frac{1}{c^2}\frac{\partial \mathbf E}{\partial t}=\mu_0 \mathbf J}
$$

or equivalently

$$
\boxed{\nabla\times\mathbf B=\mu_0 \mathbf J+\mu_0\epsilon_0\frac{\partial \mathbf E}{\partial t}}
$$

which is the **Ampère–Maxwell law**.

---

# 3. The homogeneous tensor equation

Now take

$$
\partial_{[\alpha}F_{\beta\gamma]}=0.
$$

Expanded, that means

$$
\partial_\alpha F_{\beta\gamma}
+\partial_\beta F_{\gamma\alpha}
+\partial_\gamma F_{\alpha\beta}=0.
$$

This also contains four independent equations.

---

## All spatial indices: $(\alpha,\beta,\gamma)=(1,2,3)$

Then

$$
\partial_1 F_{23}+\partial_2 F_{31}+\partial_3 F_{12}=0.
$$

Using the magnetic components of $F$,

$$
F_{23}=B_x,\quad F_{31}=B_y,\quad F_{12}=B_z
$$

up to the same convention, we get

$$
\frac{\partial B_x}{\partial x}
+\frac{\partial B_y}{\partial y}
+\frac{\partial B_z}{\partial z}=0.
$$

So

$$
\boxed{\nabla\cdot \mathbf B=0}
$$

which is **Gauss’s law for magnetism**.

---

## One time index and two spatial indices

Take $(\alpha,\beta,\gamma)=(0,1,2)$:

$$
\partial_0 F_{12}+\partial_1 F_{20}+\partial_2 F_{01}=0.
$$

Substituting the field components gives one component of Faraday’s law. Repeating for the other cyclic choices yields

$$
\boxed{\nabla\times \mathbf E=-\frac{\partial \mathbf B}{\partial t}}
$$

which is **Faraday’s law**.

---

# 4. Summary of how the four equations arise

From

$$
\partial_\mu F^{\mu\nu}=\mu_0 J^\nu
$$

you get:

* $\nu=0 \rightarrow \nabla\cdot\mathbf E=\rho/\epsilon_0$
* $\nu=1,2,3 \rightarrow \nabla\times\mathbf B=\mu_0\mathbf J+\mu_0\epsilon_0\partial_t\mathbf E$

From

$$
\partial_{[\alpha}F_{\beta\gamma]}=0
$$

you get:

* all spatial indices $\rightarrow \nabla\cdot\mathbf B=0$
* one time + two spatial indices $\rightarrow \nabla\times\mathbf E=-\partial_t\mathbf B$

So the familiar four equations are really the component form of these two tensor equations.

---

# 5. Why there are “four Maxwell equations” but only “two tensor equations”

Each tensor equation contains several component equations.

* $\partial_\mu F^{\mu\nu}=\mu_0 J^\nu$ has 4 components, but only 4 independent equations packaged as 1 scalar + 3 vector components.
* $\partial_\mu \tilde F^{\mu\nu}=0$ also has 4 components, again packaging 1 scalar + 3 vector components.

Together they reproduce the usual “four Maxwell equations”:

$$
\nabla\cdot\mathbf E=\frac{\rho}{\epsilon_0}
$$

$$
\nabla\cdot\mathbf B=0
$$

$$
\nabla\times\mathbf E=-\frac{\partial \mathbf B}{\partial t}
$$

$$
\nabla\times\mathbf B=\mu_0\mathbf J+\mu_0\epsilon_0\frac{\partial \mathbf E}{\partial t}
$$

---

# 6. The cleanest way to remember it

$$
\boxed{\partial_\mu F^{\mu\nu}=\mu_0 J^\nu}
\qquad\text{and}\qquad
\boxed{\partial_\mu \tilde F^{\mu\nu}=0}
$$

where:

* the first says **charges and currents source the field**
* the second says **the field comes from a potential and has no magnetic monopole source**

---

# 7. One subtle point

If someone says “the single Maxwell equation,” they may be speaking loosely, but in standard tensor notation there is no single equation involving only $F_{\mu\nu}$ that replaces all four classical equations without introducing extra structure. The standard relativistic packaging is these **two** equations.
