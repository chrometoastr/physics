### Calculus of Variations

## Core objects
- Generalized coordinates: $q^i(t)$
- Lagrangian: $L(q, \dot q, t)$
- Action: $S[q] = \int_{t_1}^{t_2} L(q,\dot q,t)\,dt$

## Principle of stationary action
$\delta S = 0$ (with fixed endpoints $\delta q^i(t_1)=\delta q^i(t_2)=0)$
⇒ Euler–Lagrange:
$\frac{d}{dt}\left(\frac{\partial L}{\partial \dot q^i}\right) - \frac{\partial L}{\partial q^i} = 0$

```python
# Suggested imports (keep it consistent across notebooks)
import sympy as sp
sp.init_printing()

import numpy as np
```

## Example: Harmonic oscillator from the action
$$
L = \frac{1}{2}m\dot x^2 - \frac{1}{2}k x^2
E–L ⇒ m\ddot x + kx = 0
$$

```python
t = sp.symbols('t', real=True)
m, k = sp.symbols('m k', positive=True, real=True)
x = sp.Function('x')(t)

xd = sp.diff(x, t)
L = sp.Rational(1,2)*m*xd**2 - sp.Rational(1,2)*k*x**2

EL = sp.diff(sp.diff(L, xd), t) - sp.diff(L, x)
sp.simplify(EL)
```

# 1. What is the “Action”?

At a high level:

> **The action is a single number assigned to an entire path (history) of a system.**

Instead of asking:

* “What is the force at this instant?” (Newton)

We ask:

* **“Which entire path does nature choose?”**

---

## A concrete example (particle in 1D)

A particle moves from point A at time (t_1) to point B at time (t_2).

There are infinitely many possible paths:

![Image](https://i.sstatic.net/Mdr9m.jpg)

![Image](https://encyclopedia.pub/media/common/202209/mceclip1-6332ca2ea6a6a.png)

![Image](https://encyclopedia.pub/media/common/202210/imagetools0-63491bd3cd279.png)

![Image](https://www.mdpi.com/aerospace/aerospace-11-00250/article_deploy/html/images/aerospace-11-00250-g001-550.jpg)

Nature chooses **one special path**.

---

## The rule

We define the **action**:

$$

S[q] = \int_{t_1}^{t_2} L(q, \dot{q}, t), dt

$$

Where:

* (q(t)): the path
* (\dot{q}): velocity
* (L): the **Lagrangian**

---

## The Lagrangian

For most systems:

$$
L = T - V
$$

* (T): kinetic energy
* (V): potential energy

---

## Interpretation

The action is:

> **A score assigned to a whole trajectory**

Nature chooses the path where:

> **The action is stationary (usually minimized)**

---

# 2. What does “varying the action” mean?

This is the heart of the idea.

---

## Think like this

You start with the “true” path (q(t)).

Now imagine slightly modifying it:

$$
q(t) \rightarrow q(t) + \epsilon,\eta(t)
$$

Where:

* $\epsilon$: tiny number
* $\eta(t)$: arbitrary small deformation
* endpoints fixed: $\eta(t_1) = \eta(t_2) = 0$

---

## Visual intuition

![Image](https://farside.ph.utexas.edu/teaching/336k/Newtonhtml/img1719.png)

![Image](https://media.springernature.com/m685/springer-static/image/art%3A10.1038%2Fs41592-024-02378-4/MediaObjects/41592_2024_2378_Fig1_HTML.png)

![Image](https://openoregon.pressbooks.pub/app/uploads/sites/210/2018/08/Stress_Strain.jpg)

![Image](https://i.sstatic.net/SYTKk.png)

You are comparing:

* the real path
* nearby “almost” paths

---

## Varying the action

Plug the modified path into the action:

$$
S[q + \epsilon \eta]
$$

Then compute how it changes:

$$
\delta S = \left.\frac{dS}{d\epsilon}\right|_{\epsilon=0}
$$

---

## Principle of stationary action

$$
\delta S = 0
$$

This means:

> Small changes in the path do **not** change the action (to first order)

---

# 3. What do you get from this?

When you carry out the math, you get:

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}}\right) - \frac{\partial L}{\partial q} = 0
$$

This is the **Euler–Lagrange equation**.

---

## Key insight

> **All of classical mechanics comes from this one condition:**
>
> $$
> \delta S = 0
> $$

Newton’s laws are **derived**, not fundamental.

---

# 4. Why this is powerful

This idea generalizes beautifully.

---

## A. Works for any coordinates

You don’t need forces—just energies.

---

## B. Works for fields

Instead of $q(t)$, you have a field:

$$
\phi(x, t)
$$

Action becomes:

$$
S[\phi] = \int \mathcal{L}(\phi, \partial_\mu \phi), d^4x
$$

---

## C. Leads to conservation laws

Symmetries of the action → conserved quantities
(**Noether’s theorem**, which your study plan highlights )

---

## D. It’s the foundation of GR

This is where things get really interesting.

---

# 5. Action in General Relativity

Instead of a particle path, we now consider:

> **The geometry of spacetime itself**

---

## The Einstein–Hilbert action

$$
S = \int R \sqrt{-g}, d^4x + S_{\text{matter}}
$$

Where:

* $R$: Ricci scalar (curvature)
* $g$: determinant of metric
* $S_{\text{matter}}$: matter fields

---

## What do we vary?

We vary the **metric**:

$$
g_{\mu\nu} \rightarrow g_{\mu\nu} + \delta g_{\mu\nu}
$$

---

## Result

Setting:

$$
\delta S = 0
$$

gives:

$$
G_{\mu\nu} = 8\pi T_{\mu\nu}
$$

👉 Einstein’s field equations.

---

## Deep insight

> **GR is just “vary the action” applied to spacetime itself**

# 7. Intuition upgrade (important)

A subtle but critical point:

> The action is not necessarily **minimized**

It is:

* minimized
* maximized
* or saddle-point

The correct statement is:

> **The action is stationary**
